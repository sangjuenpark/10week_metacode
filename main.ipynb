# Git 과제 - Step1: 라이브러리 불러오기 및 데이터 읽기

```python
# 데이터 읽기용 코드셀입니다.

import pandas as pd
print("="*20, "game.xlsx 데이터 정보입니다.", "="*20)
tmp_df = pd.read_excel("./game.xlsx", sheet_name="Sheet1")
display(tmp_df.shape)
display(tmp_df.columns)
display(tmp_df.head())
```

# Git 과제 - Step2: 데이터 전처리

```python
# features와 target으로 데이터 분리
features = tmp_df[["playtime", "playcount", "level", "paycount", "payamount", "mainjoblevel"]]
target = tmp_df["mainjob"]

# features 정보 출력
print("="*20, "features 정보입니다.", "="*20)
display(features.shape)
display(features.head())

# target 정보 출력  
print("="*20, "target 정보입니다.", "="*20)
display(target.shape)
display(target.head())
```

```python
from sklearn.model_selection import train_test_split

# features와 target 데이터를 train_test_split을 통해 분리
X_train, X_test, y_train, y_test = train_test_split(
    features, target, 
    test_size=0.2, 
    random_state=42
)

# 분할 이후 train 데이터 크기 출력
print("="*20, "분할 이후 train 데이터 크기입니다.", "="*20)
print(X_train.shape)
print(y_train.shape)

# 분할 이후 test 데이터 크기 출력
print("="*20, "분할 이후 test 데이터 크기입니다.", "="*20)
print(X_test.shape)
print(y_test.shape)

# train 데이터와 test 데이터 비율 계산 결과 출력
print("="*20, "train 데이터와 test 데이터 비율 계산 결과입니다.", "="*20)
print(len(X_train) / len(features))  # train 비율
print(len(X_test) / len(features))   # test 비율
```

```python
from sklearn.preprocessing import LabelEncoder

# LabelEncoder 객체 생성
le = LabelEncoder()

# target 데이터에 LabelEncoding 적용
y_train_encoded = le.fit_transform(y_train)
y_test_encoded = le.transform(y_test)

# LabelEncoding 결과 확인
print("="*20, "LabelEncoding 전후 비교입니다.", "="*20)
print("원본 target 데이터 (처음 10개):")
print(y_train.head(10).values)
print("
LabelEncoding 후 데이터 (처음 10개):")
print(y_train_encoded[:10])

# 클래스 매핑 정보 확인
print("
클래스 매핑 정보:")
for i, class_name in enumerate(le.classes_):
    print(f"{class_name} -> {i}")

# 인코딩된 데이터 형태 확인
print("
인코딩된 데이터 형태:")
print(f"y_train_encoded shape: {y_train_encoded.shape}")
print(f"y_test_encoded shape: {y_test_encoded.shape}")
```

# Git 과제 - Step3: 모델 학습

```python
from interpret.glassbox import ExplainableBoostingClassifier
from sklearn.metrics import accuracy_score, classification_report

# ExplainableBoostingClassifier 모델 생성 및 학습
ebm = ExplainableBoostingClassifier(random_state=42)
ebm.fit(X_train, y_train_encoded)

# 예측 수행
y_pred = ebm.predict(X_test)

# 정확도 계산 및 출력
accuracy = accuracy_score(y_test_encoded, y_pred)
print("="*20, "accuracy 결과입니다.", "="*20)
print(accuracy)

# 분류 보고서 생성 및 출력
print("="*20, "classification_report 결과입니다.", "="*20)
# 원본 클래스명으로 변환하여 보고서 생성
y_test_original = le.inverse_transform(y_test_encoded)
y_pred_original = le.inverse_transform(y_pred)
print(classification_report(y_test_original, y_pred_original, target_names=["dealer", "healer", "tanker"]))
```
